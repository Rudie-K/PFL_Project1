## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
##### Constants

# Bases allowed in a valid read (after converting to upper case)
VALID_BASES = "ACGT"


##### Functions

def validate_read(sequence, reference):
    """
    Normalise a sequencing read to upper case and check it is valid.

    A read is valid only if it is the same length as the reference and
    every character is one of A, C, G or T (after upper-casing).

    Args:
        sequence: The read's nucleotide sequence (any length/contents).
        reference: The reference sequence (non-empty, upper-case A/C/G/T).

    Returns:
        A dictionary with keys:
            "sequence": the upper-case version of the read (even if invalid)
            "valid": True if the read passed all checks, otherwise False
            "errors": list of error codes, in the order "wrong_length",
                      then "invalid_characters"; [] for a valid read
    """
    normalised_sequence = sequence.upper()
    errors = []

    # Length is checked on the original read, because upper-casing some
    # non-DNA characters changes the length (e.g. "ß".upper() == "SS")
    if len(sequence) != len(reference):
        errors.append("wrong_length")

    # Any character outside A/C/G/T (e.g. N, U, "-", spaces) is invalid
    if not all(base in VALID_BASES for base in normalised_sequence):
        errors.append("invalid_characters")

    return {
        "sequence": normalised_sequence,
        "valid": len(errors) == 0,
        "errors": errors,
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """
    Find every position where a valid read differs from the reference.

    Args:
        sequence: A valid read (upper-case A/C/G/T, same length as reference).
        reference: The reference sequence (upper-case A/C/G/T).

    Returns:
        A list with one dictionary per mismatch, in ascending position order.
        Each dictionary has the keys:
            "position": position of the mismatch, numbered from 1
            "reference_base": the base in the reference at that position
            "read_base": the base in the read at that position
        Returns [] if the read matches the reference exactly.
    """
    mutations = []

    # Compare the read and reference base-by-base; looping from the start
    # means mismatches are recorded in ascending position order
    for index in range(len(sequence)):
        if sequence[index] != reference[index]:
            mutations.append({
                "position": index + 1,  # convert Python's 0-based index to 1-based position
                "reference_base": reference[index],
                "read_base": sequence[index],
            })

    return mutations
````

````python
# Test find_mutations with normal inputs and edge cases
# (inputs can be assumed valid, so invalid inputs are not tested here)

##### Normal inputs
assert find_mutations("GTGA", "ATGC") == [{"position": 1, "reference_base": "A", "read_base": "G"},
                                          {"position": 4, "reference_base": "C", "read_base": "A"}], \
    f"Example from the brief should give mismatches at positions 1 and 4, got {find_mutations('GTGA', 'ATGC')}"
assert find_mutations("ATTC", "ATGC") == [{"position": 3, "reference_base": "G", "read_base": "T"}], \
    f"Should find one mismatch at position 3, got {find_mutations('ATTC', 'ATGC')}"

##### Edge cases
assert find_mutations("ATGC", "ATGC") == [], \
    f"A perfect match should give [], got {find_mutations('ATGC', 'ATGC')}"
assert find_mutations("GTGC", "ATGC") == [{"position": 1, "reference_base": "A", "read_base": "G"}], \
    f"First position should be numbered 1, not 0, got {find_mutations('GTGC', 'ATGC')}"
assert find_mutations("ATGA", "ATGC") == [{"position": 4, "reference_base": "C", "read_base": "A"}], \
    f"A mismatch at the final position should be found, got {find_mutations('ATGA', 'ATGC')}"
assert find_mutations("TACG", "ATGC") == [{"position": 1, "reference_base": "A", "read_base": "T"},
                                          {"position": 2, "reference_base": "T", "read_base": "A"},
                                          {"position": 3, "reference_base": "G", "read_base": "C"},
                                          {"position": 4, "reference_base": "C", "read_base": "G"}], \
    f"Every position differs: all 4 should be listed in ascending order, got {find_mutations('TACG', 'ATGC')}"
assert find_mutations("G", "A") == [{"position": 1, "reference_base": "A", "read_base": "G"}], \
    f"One-base sequences should work, got {find_mutations('G', 'A')}"
assert find_mutations("", "") == [], \
    f"Empty strings have nothing to compare, so should give [], got {find_mutations('', '')}"

print("All find_mutations tests passed")

##### Check on the real dataset
n_events = 0
n_mutated_reads = 0
for read in reads:
    read_mutations = find_mutations(validate_read(read["sequence"], reference)["sequence"], reference)
    n_events += len(read_mutations)
    if read_mutations:
        n_mutated_reads += 1
print(f"Provided dataset: {n_mutated_reads} reads with at least one mismatch; {n_events} mismatch events in total")
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """
    Validate every read and find the mismatches in each valid read.

    Uses validate_read (Step 1) and find_mutations (Step 2). A new list of
    new dictionaries is built, so the original read dictionaries are not
    changed.

    Args:
        reads: List of read dictionaries with keys "read_id", "sample",
               "group" and "sequence" (all strings).
        reference: The reference sequence (non-empty, upper-case A/C/G/T).

    Returns:
        A list with one dictionary per read, in the same order as the input,
        with keys:
            "read_id", "sample", "group": copied from the input read
            "sequence": the upper-case version of the read
            "status": "valid" or "invalid"
            "errors": list of validation error codes ([] if valid)
            "mutations": list of mismatch dictionaries from find_mutations
                         ([] if the read is invalid or matches the reference)
        Returns [] if reads is empty.
    """
    analysed_reads = []

    for read in reads:
        validation = validate_read(read["sequence"], reference)

        # Only valid reads can be compared base-by-base with the reference
        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(validation["sequence"], reference)
        else:
            status = "invalid"
            mutations = []

        # Build a new dictionary rather than editing the original read
        analysed_reads.append({
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": validation["errors"],
            "mutations": mutations,
        })

    return analysed_reads
````

````python
##### Constants

# Number of most-mutated positions to report for each sample
N_TOP_POSITIONS = 3

# Text shown in tables when a percentage cannot be calculated, and its explanation
NOT_AVAILABLE = "n/a"
NOT_AVAILABLE_NOTE = f"{NOT_AVAILABLE} = percentage cannot be calculated (sample has no valid reads)"


##### Helper functions

def calculate_percentage(count, total):
    """Return count as a percentage of total, or None if total is 0."""
    if total == 0:
        return None
    return count / total * 100


def find_top_positions(position_counts, n_positions=N_TOP_POSITIONS):
    """
    Return the positions with the most mismatch events, highest first.

    Positions with a count of 0 are ignored, so fewer than n_positions are
    returned if fewer positions have mismatches. Ties are broken by choosing
    the earlier position.

    Args:
        position_counts: Dictionary of {position: number of mismatch events}.
        n_positions: Maximum number of positions to return.

    Returns:
        A list of up to n_positions positions (integers).
    """
    # Sort by position first; Python's sort keeps the existing order of
    # tied items, so after sorting by count, tied positions stay earliest-first
    by_position = sorted(position_counts.items())
    by_count = sorted(by_position, key=lambda item: item[1], reverse=True)

    top_positions = []
    for position, count in by_count:
        if count > 0 and len(top_positions) < n_positions:
            top_positions.append(position)
    return top_positions


def group_reads_by_sample(analysed_reads):
    """Return a dictionary of {sample name: list of its analysed reads}, in first-seen order."""
    reads_by_sample = {}
    for read in analysed_reads:
        sample = read["sample"]
        if sample not in reads_by_sample:
            reads_by_sample[sample] = []
        reads_by_sample[sample].append(read)
    return reads_by_sample


def summarise_sample(sample_reads):
    """
    Count valid/invalid reads, mismatches, substitutions and positions for one sample.

    Args:
        sample_reads: Non-empty list of analysed reads (from analyse_reads)
                      that all come from the same sample.

    Returns:
        A dictionary in the per-sample format required by summarise_samples.
    """
    valid_reads = 0
    invalid_reads = 0
    mismatch_reads = 0
    mismatch_events = 0
    substitution_counts = {}
    position_counts = {}

    for read in sample_reads:
        if read["status"] == "valid":
            valid_reads += 1
            if read["mutations"]:
                mismatch_reads += 1  # a read counts once, however many mismatches it has

            for mutation in read["mutations"]:
                mismatch_events += 1  # every mismatch counts as one event

                substitution = f'{mutation["reference_base"]}>{mutation["read_base"]}'
                substitution_counts[substitution] = substitution_counts.get(substitution, 0) + 1

                position = mutation["position"]
                position_counts[position] = position_counts.get(position, 0) + 1
        else:
            invalid_reads += 1

    return {
        "group": sample_reads[0]["group"],
        "valid_reads": valid_reads,
        "invalid_reads": invalid_reads,
        "mismatch_reads": mismatch_reads,
        # Percentage uses valid reads only; None if the sample has no valid reads
        "mismatch_percentage": calculate_percentage(mismatch_reads, valid_reads),
        "mismatch_events": mismatch_events,
        # Sorted so the dictionaries are easy to read when printed
        "substitution_counts": dict(sorted(substitution_counts.items())),
        "position_counts": dict(sorted(position_counts.items())),
        "top_positions": find_top_positions(position_counts),
    }


def format_percentage(percentage):
    """Return a percentage as text to 1 decimal place, or NOT_AVAILABLE if it is None."""
    if percentage is None:
        return NOT_AVAILABLE
    return f"{percentage:.1f}"


def print_results_table(sample_summary, sample_names):
    """Print the table of summarised results, with one row per sample."""
    header = f'{"Sample":<12}{"Valid reads":>13}{"Invalid reads":>15}'
    header += f'{"% with mismatch":>17}{"Mismatch events":>17}   Top 3 positions'
    print(header)
    print("-" * len(header))

    for sample in sample_names:
        summary = sample_summary[sample]
        if summary["top_positions"]:
            top_positions_text = ", ".join([str(position) for position in summary["top_positions"]])
        else:
            top_positions_text = "none"

        row = f'{sample:<12}{summary["valid_reads"]:>13}{summary["invalid_reads"]:>15}'
        row += f'{format_percentage(summary["mismatch_percentage"]):>17}'
        row += f'{summary["mismatch_events"]:>17}   {top_positions_text}'
        print(row)


def print_substitution_table(sample_summary, sample_names):
    """Print substitution counts and rates (per 100 valid reads), with samples as columns."""
    # Collect every substitution seen in any sample in this group
    substitutions = []
    for sample in sample_names:
        for substitution in sample_summary[sample]["substitution_counts"]:
            if substitution not in substitutions:
                substitutions.append(substitution)

    if not substitutions:
        print("No substitutions observed.")
    else:
        header = f'{"Substitution":<12}'
        for sample in sample_names:
            header += f"{sample:>16}"
        print(header)
        print("-" * len(header))

        for substitution in sorted(substitutions):
            # Show "A>G" as "A → G" for readability (the returned dictionary keeps "A>G")
            substitution_label = substitution.replace(">", " → ")
            row = f"{substitution_label:<12}"
            for sample in sample_names:
                summary = sample_summary[sample]
                count = summary["substitution_counts"].get(substitution, 0)
                rate = calculate_percentage(count, summary["valid_reads"])
                cell_text = f"{count} ({format_percentage(rate)})"
                row += f"{cell_text:>16}"
            print(row)


def print_group_summaries(sample_summary):
    """Print a results table and a substitution table for each experimental group."""
    if not sample_summary:
        print("No reads to summarise.")
    else:
        # List the samples in each group, keeping first-seen order
        samples_by_group = {}
        for sample, summary in sample_summary.items():
            group = summary["group"]
            if group not in samples_by_group:
                samples_by_group[group] = []
            samples_by_group[group].append(sample)

        for group, sample_names in samples_by_group.items():
            print(f"\n===== Group: {group} =====\n")
            print("Summary of results (percentages use valid reads only)")
            print_results_table(sample_summary, sample_names)
            print("\nSubstitutions: count (per 100 valid reads)")
            print_substitution_table(sample_summary, sample_names)

            # Explain "n/a" if any sample in this group has no valid reads
            for sample in sample_names:
                if sample_summary[sample]["valid_reads"] == 0:
                    print(f"\n{NOT_AVAILABLE_NOTE}")
                    break


##### Main function

def summarise_samples(analysed_reads):
    """
    Summarise mismatch patterns for each sample and print a summary per group.

    The input list is not changed. Percentages use valid reads only.

    Args:
        analysed_reads: List of analysed read dictionaries from analyse_reads.

    Returns:
        A dictionary of {sample name: per-sample summary dictionary} with keys
        "group", "valid_reads", "invalid_reads", "mismatch_reads",
        "mismatch_percentage" (None if no valid reads), "mismatch_events",
        "substitution_counts", "position_counts" and "top_positions".
        Returns {} if analysed_reads is empty.
    """
    sample_summary = {}
    reads_by_sample = group_reads_by_sample(analysed_reads)

    for sample, sample_reads in reads_by_sample.items():
        sample_summary[sample] = summarise_sample(sample_reads)

    print_group_summaries(sample_summary)
    return sample_summary
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:

### 👍 You can assume

- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

````python
##### Helper functions

def calculate_mean(values):
    """Return the mean of a list of numbers, or None if the list is empty."""
    if len(values) == 0:
        return None
    return sum(values) / len(values)


def combine_position_counts(sample_summary, groups):
    """
    Add up the mismatch events at every position across all samples in the given groups.

    Args:
        sample_summary: Dictionary returned by summarise_samples.
        groups: List of group names whose samples should be combined.

    Returns:
        A list of combined counts, where index 0 holds position 1, index 1
        holds position 2, and so on. The list runs up to the highest position
        with a mismatch, because the reference length is not passed to
        compare_positions; any later positions have no mismatches.
    """
    # Find the highest position with a mismatch in the selected samples
    highest_position = 0
    for summary in sample_summary.values():
        if summary["group"] in groups:
            for position in summary["position_counts"]:
                if position > highest_position:
                    highest_position = position

    # Start with a count of 0 at every position
    combined_counts = []
    for index in range(highest_position):
        combined_counts.append(0)

    # Add each sample's counts (position 1 is stored at index 0)
    for summary in sample_summary.values():
        if summary["group"] in groups:
            for position, count in summary["position_counts"].items():
                combined_counts[position - 1] += count

    return combined_counts


def get_position_percentages(sample_summary, group, position):
    """
    Return the percentage of valid reads with a mismatch at position, for each sample in group.

    Samples with no valid reads are skipped, because their percentage cannot
    be calculated. A sample with no mismatch at the position gives 0.0.
    """
    percentages = []
    for summary in sample_summary.values():
        if summary["group"] == group and summary["valid_reads"] > 0:
            # Each read can have at most one mismatch at a position, so the
            # number of events there equals the number of reads with a mismatch there
            count = summary["position_counts"].get(position, 0)
            percentages.append(calculate_percentage(count, summary["valid_reads"]))
    return percentages


def find_excluded_samples(sample_summary, groups):
    """Return the names of samples in the given groups that have no valid reads."""
    excluded_samples = []
    for sample, summary in sample_summary.items():
        if summary["group"] in groups and summary["valid_reads"] == 0:
            excluded_samples.append(sample)
    return excluded_samples


def format_value(value):
    """Return a number as text to 2 decimal places, or NOT_AVAILABLE if it is None."""
    if value is None:
        return NOT_AVAILABLE
    return f"{value:.2f}"


def format_p_value(p_value):
    """Return a p-value as text in standard style, e.g. 'p = 0.012' or 'p < 0.001'."""
    if p_value is None:
        return f"p = {NOT_AVAILABLE}"
    if p_value < 0.001:
        return "p < 0.001"
    return f"p = {p_value:.3f}"


def print_position_comparison(comparison, control_group, treated_group):
    """Print one line summarising the statistical comparison at one position."""
    # 2 decimal places, so that small control rates (e.g. 0.03%) are not shown as 0.00
    treated_mean_text = format_value(comparison["treated_mean"])
    control_mean_text = format_value(comparison["control_mean"])

    if comparison["difference"] is None:
        difference_text = NOT_AVAILABLE
    else:
        difference_text = f"{comparison['difference']:+.2f}"  # + or - sign shows the direction

    if comparison["statistic"] is None:
        statistics_text = "t-test not possible (needs 2+ samples per group, with some variation)"
    else:
        statistics_text = f"t = {comparison['statistic']:.2f}, {format_p_value(comparison['p_value'])}"

    line = f"Position {comparison['position']}: "
    line += f"{treated_group} {treated_mean_text}% (n = {comparison['treated_n']}) vs "
    line += f"{control_group} {control_mean_text}% (n = {comparison['control_n']}); "
    line += f"difference {difference_text} percentage points; {statistics_text}"
    print(line)


##### Main function

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """
    Statistically compare mismatch rates between two groups at the top mismatch positions.

    Combines the position counts from both groups, selects the top positions
    (earlier position wins ties), calculates each sample's percentage of valid
    reads with a mismatch at each position, and compares the groups with a
    two-sided Welch's t-test (calculate_statistics). Samples with no valid
    reads are excluded and reported. Results are printed, one line per position.

    Args:
        sample_summary: Dictionary returned by summarise_samples.
        control_group: Name of the control group.
        treated_group: Name of the treated group.

    Returns:
        A list with one dictionary per selected position, with keys
        "position", "control_n", "treated_n", "control_mean", "treated_mean",
        "difference" (treated mean - control mean, in percentage points),
        "statistic" and "p_value". Values that cannot be calculated are None.
        Returns [] if no mismatches are found in either group.
    """
    groups = [control_group, treated_group]

    # Report samples that cannot contribute a percentage
    for sample in find_excluded_samples(sample_summary, groups):
        print(f"Excluded {sample}: no valid reads, so its mismatch percentages cannot be calculated")

    # Combine both groups' counts, then convert to {position: count} so that
    # find_top_positions (Step 4) can select the top positions with the same tie rule
    combined_counts = combine_position_counts(sample_summary, groups)
    combined_position_counts = {}
    for index in range(len(combined_counts)):
        combined_position_counts[index + 1] = combined_counts[index]
    top_positions = find_top_positions(combined_position_counts)

    if not top_positions:
        print(f"No mismatches found in the {control_group} or {treated_group} groups, so no positions to compare.")
        return []

    print(f"Two-sided Welch's t-test of per-sample mismatch percentages ({treated_group} vs {control_group}):")

    comparisons = []
    for position in top_positions:
        control_percentages = get_position_percentages(sample_summary, control_group, position)
        treated_percentages = get_position_percentages(sample_summary, treated_group, position)
        control_mean = calculate_mean(control_percentages)
        treated_mean = calculate_mean(treated_percentages)

        if control_mean is None or treated_mean is None:
            difference = None
        else:
            difference = treated_mean - control_mean

        statistical_result = calculate_statistics(control_percentages, treated_percentages)

        comparison = {
            "position": position,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistical_result["statistic"],
            "p_value": statistical_result["p_value"],
        }
        comparisons.append(comparison)
        print_position_comparison(comparison, control_group, treated_group)

    return comparisons
````
